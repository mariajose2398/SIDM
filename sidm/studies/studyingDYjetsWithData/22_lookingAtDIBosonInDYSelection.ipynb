```python
import os, sys
sidm_path = str(os.getcwd()).split('/sidm')[0]
if sidm_path not in sys.path:
    sys.path.insert(1, sidm_path)
from sidm.tools import utilities #scaleout, sidm_processor, llpnanoaodschema
from sidm.studies.cosmicVeto import functions
utilities.set_plot_style()
```

```python
DB = functions.get_bg_file_names("Diboson")
DY = functions.get_bg_file_names("DYJetsToLL")
channel_name  = "cr_Zmumu_noLJ"
folder_name = f"OutputFiles/{channel_name}"
summed_DB  = utilities.sum_hist(DB,  folder_name)
summed_DY  = utilities.sum_hist(DY, folder_name)
```

```python
histogram_list = list(summed_DB.keys())
```

```python
for x in histogram_list:
    print(x)
```

```python
from matplotlib.colors import LogNorm
import matplotlib.pyplot as plt
for x in histogram_list:
    print(x)
    # fig, axs =  plt.subplots(1, 2, figsize=(30, 12))
    
    try:
        plt.figure()
        histogram = summed_DB [x][channel_name, :]
        utilities.plot(histogram, label ="DB")
        histogram = summed_DY [x][channel_name, :]
        utilities.plot(histogram, label ="DY")
        # histogram = sum_data_all [x][channel_name, :]
        # utilities.plot(histogram, ax=axs[1], skip_label=True)
        plt.yscale("log")
        plt.legend()
        plt.show()
    except:
        # continue
        n=2
        fig, axs  = plt.subplots(1, n, figsize=(n*12, 10))
        histogram = summed_DB [x][channel_name, :, :]
        utilities.plot(histogram, norm=LogNorm(1e1, 1e5), label="DB", ax = axs[0])
        axs[0].set_title("DB")
        histogram = summed_DY [x][channel_name, :, :]
        utilities.plot(histogram, norm=LogNorm(1e1, 1e5), label="DB", ax=axs[1])
        axs[1].set_title("DY")
        # histogram = sum_data_all [x][channel_name, :, :]
        # utilities.plot(histogram, ax=axs[1], skip_label=True, norm=LogNorm(1e1, 1e5))
        for i in range(n):
            axs[i].legend()
        plt.show()
    
        
```
